```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
logs=pd.read_csv('../data/cleaned_logs.csv')
```

```python
X=logs["template"]
```

```python
X
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X_vectorized = vectorizer.fit_transform(X)
```

```python
X_vectorized_df=pd.DataFrame(X_vectorized.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
X_vectorized_df.head()
```

```python
# 1) Rows where message == template
exact_match_mask = logs["message"] == logs["template"]
exact_match_rows = logs[exact_match_mask]
exact_match_count = exact_match_mask.sum()
exact_match_pct = exact_match_count / len(logs) * 100

print(f"Rows where message == template: {exact_match_count}/{len(logs)} ({exact_match_pct:.2f}%)")
display(exact_match_rows.head(10)[["message", "template", "cluster_id"]])

# 2) Rows where template contains "<*>"
generalized_mask = logs["template"].str.contains("<*>", regex=False, na=False)
generalized_rows = logs[generalized_mask]
generalized_count = generalized_mask.sum()
generalized_pct = generalized_count / len(logs) * 100

print(f"\nRows where template contains '<*>': {generalized_count}/{len(logs)} ({generalized_pct:.2f}%)")
display(generalized_rows.head(10)[["message", "template", "cluster_id"]])

# 3) Top 20 most frequent templates
top20_templates = logs["template"].value_counts().head(20).rename_axis("template").reset_index(name="count")
print("\nTop 20 most frequent templates:")
display(top20_templates)

plt.figure(figsize=(12, 6))
sns.barplot(data=top20_templates, x="count", y="template", palette="viridis")
plt.title("Top 20 Most Frequent Templates")
plt.xlabel("Count")
plt.ylabel("Template")
plt.tight_layout()
plt.show()
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
print(model)
```

```python
embeddings = model.encode(
    logs["template"].tolist(),
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
embeddings.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = scaler.fit_transform(embeddings)
```

```python
transformer_iforest = IsolationForest(random_state=42, n_estimators=200, n_jobs=-1)
transformer_iforest.fit(X)
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2, random_state=42)

pca_result = pca.fit_transform(embeddings)

logs["PC1"] = pca_result[:,0]
logs["PC2"] = pca_result[:,1]

plt.figure(figsize=(10,8))

plt.scatter(
    logs["PC1"],
    logs["PC2"],
    c=logs["anomaly"],
    cmap="coolwarm",
    alpha=0.7
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Isolation Forest Anomalies in Sentence Transformer Embedding Space")

plt.colorbar(label="Prediction (1=Normal, -1=Anomaly)")
plt.show()
```

```python
import joblib
from pathlib import Path

SAVE_DIR = Path("ml/Training/models/logs")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

artifacts = {}

if 'tfidf_iforest' in locals():
    artifacts['tfidf_iforest'] = tfidf_iforest
if 'transformer_iforest' in locals():
    artifacts['transformer_iforest'] = transformer_iforest
if 'vectorizer' in locals():
    artifacts['tfidf_vectorizer'] = vectorizer
if 'scaler' in locals():
    artifacts['scaler'] = scaler
if 'pca' in locals():
    artifacts['pca'] = pca

for name, obj in artifacts.items():
    joblib.dump(obj, SAVE_DIR / f"{name}.pkl")

logs.to_csv(SAVE_DIR / "logs_with_transformer_anomalies.csv", index=False)

print("✓ models saved")
print("✓ processed dataframe exported")
print("✓ success messages")
```
